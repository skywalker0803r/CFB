預測目標:DeSOx_1st 其餘為特徵

1. 先將數據select_df依據時間排序。

2. 自第1001筆開始，取用前1000筆數據建立rf模型，評估建模數據的"預測誤差絕對值"，計算前述數值的第80百分位數當成"門檻值"。

3. 運用2.的模型執行第1001筆的預測。

4. 比對3.的結果之"預測誤差絕對值"。若此數值超過"門檻值"，則第1002筆預測重複2.＆3.步驟；若此數值不超過"門檻值"，直接運用2.的模型執行第1002筆的預測。

5. 之後以此類推....

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from tqdm import tqdm
import warnings
warnings.filterwarnings("ignore")

# ======================== 使用者自定義超參數 ==========================
PERCENTILE_THRESHOLD = 95     # 門檻百分位
PREDICTION_INTERVAL = 300     # 每次預測區間 (筆數)
MAX_STEPS = 1000              # 最多預測幾次（非總筆數）
WINDOW_SIZE = 1000            # 每次用多少資料訓練
# =====================================================================

# 1. 資料排序
select_df = select_df.sort_index().reset_index(drop=True)

# 2. 設定欄位
target_col = y_col
feature_cols = [col for col in select_df.columns if col not in ['timestamp', target_col]]
start_idx = WINDOW_SIZE

# 3. 儲存預測結果
predictions = []
abs_errors = []
thresholds = []
indices = []

# 4. 初始化進度條
pbar = tqdm(total=MAX_STEPS)

# 5. 初始模型與門檻
model = None
cached_threshold = None

# 6. 開始迴圈
i = start_idx
while i + PREDICTION_INTERVAL <= len(select_df):
    if pbar.n >= MAX_STEPS:
        print(f"已達 {MAX_STEPS} 步，提前結束迴圈")
        break

    retrain_model = False

    # 如果還沒模型，或是前次預測誤差超門檻 → 重訓
    if model is None:
        retrain_model = True
    else:
        # 快速預測一整段並計算誤差
        test_df = select_df.iloc[i:i+PREDICTION_INTERVAL]
        test_X = test_df[feature_cols]
        true_y = test_df[target_col].values
        preds = model.predict(test_X)
        errors = np.abs(preds - true_y)

        # 如果任一筆超門檻 → 重訓
        if cached_threshold is not None and np.any(errors > cached_threshold):
            retrain_model = True

    if retrain_model:
        train_df = select_df.iloc[i - WINDOW_SIZE:i]
        train_X = train_df[feature_cols]
        train_y = train_df[target_col]

        model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
        model.fit(train_X, train_y)

        train_preds = model.predict(train_X)
        train_errors = np.abs(train_preds - train_y)
        cached_threshold = np.percentile(train_errors, PERCENTILE_THRESHOLD)

        # 預測下一段（剛好可以重用 test_df）
        test_df = select_df.iloc[i:i+PREDICTION_INTERVAL]
        test_X = test_df[feature_cols]
        true_y = test_df[target_col].values
        preds = model.predict(test_X)
        errors = np.abs(preds - true_y)

    # 紀錄整段結果
    idx_range = list(range(i, i + PREDICTION_INTERVAL))
    predictions.extend(preds.tolist())
    abs_errors.extend(errors.tolist())
    thresholds.extend([cached_threshold] * len(preds))
    indices.extend(idx_range)

    pbar.update(1)
    i += PREDICTION_INTERVAL

pbar.close()

# 7. 輸出結果表
result_df = select_df.loc[indices].copy()
result_df['prediction'] = predictions
result_df['abs_error'] = abs_errors
result_df['threshold'] = thresholds

result_df.head()
#跟我細講這段程式碼邏輯
```
